#### EDA Finding: Seasonal Pollution Spikes at Start of Year

**Observation**

The time-series plots show recurring PM2.5 spikes at the beginning of each year (January–February) across 2013–2017, despite a gradual downward trend in overall annual averages. This pattern is consistent with well-documented seasonal smog events in Beijing during this period, often referred to as part of the "Airpocalypse."

**Contributing Factors**

| Factor | Explanation |
|---|---|
| **Winter heating systems** | Municipal coal-burning heating systems across northern China ramp up at the start of winter. Startup of industrial boilers is inefficient, producing dense bursts of pollutants. |
| **Meteorological trapping** | Winter brings low pressure gradients, high humidity, and low wind speeds. Combined with Beijing's topography (mountains on three sides), stagnant air traps emissions near ground level. |
| **Lunar New Year fireworks** | Traditional fireworks at the start of the lunar calendar release large amounts of particulate matter in short bursts, often coinciding with January/February spikes. |
| **Policy and intervention effects** | Beijing's "war on pollution" (from 2014) introduced red alerts, construction halts, vehicle restrictions, and industrial bans — these interventions affected pollution levels differently year to year, adding variability to the seasonal pattern. |

**Notable historical data points**
- January 2015: PM2.5 reached ~568 µg/m³, over 20x the WHO 24-hour safety limit.
- January 2016: Following Beijing's first consecutive "red alerts" in December 2015, PM2.5 exceeded 500–700 µg/m³ in parts of northern China.

**Implications for Modeling**

- These spikes are genuine seasonal signal, not noise or data errors — they should not be removed or capped.
- `month` and `day` (and possibly an `is_winter` or `is_new_year_period` flag) may be useful engineered features, since the timing of these events is tied to the calendar (heating season onset, Lunar New Year) rather than weather conditions alone.
- Models should be evaluated on their ability to predict these high-pollution periods specifically, since this is when accurate forecasting matters most for public health warnings — a model that only performs well on "typical" low-PM2.5 hours would miss the point of the task.

*Sources: The Guardian, BBC, Smart Air Filters, Straits Times, Phys.org, ABC News, Al Jazeera, Japan Times, ScienceDirect.*

####Data Cleaning

```python
def classify_gap(duration):
    if duration < pd.Timedelta(hours=2):      # was: days=1
        return 'small'
    elif duration < pd.Timedelta(hours=8):    # was: days=3
        return 'medium'
    elif duration < pd.Timedelta(hours=24):
        return 'long'
```

```python


def find_gaps(series):
    """
    Given a datetime-indexed series (already reindexed to full hourly range),
    return a list of dicts describing each gap of missing values.
    """
    is_na = series.isna()
    gaps = []

    if not is_na.any():
        return gaps

    # Identify consecutive groups of missing values
    # group id increments every time the missing/not-missing status changes
    group_id = (is_na != is_na.shift()).cumsum()

    for _, grp in series[is_na].groupby(group_id[is_na]):
        start = grp.index[0]
        end = grp.index[-1]
        # duration = time span covered by the gap (end - start + 1 hour step)
        duration = (end - start) + pd.Timedelta(hours=1)
        gaps.append({
            'start': start,
            'end': end,
            'duration': duration,
            'severity': classify_gap(duration)
        })

    return gaps

# Main loop: for each station, reindex to full hourly range, then check every variable
gap_summary = {}  # nested dict: {station: {variable: [gap_dicts]}}

for station, sdf in station_dfs.items():
    # Build a complete hourly datetime range spanning this station's data
    full_range = pd.date_range(start=sdf.index.min(), end=sdf.index.max(), freq='H')
    sdf_full = sdf.reindex(full_range)

    gap_summary[station] = {}

    for col in sdf_full.columns:
        gaps = find_gaps(sdf_full[col])
        if gaps:  # only store variables that actually have gaps
            gap_summary[station][col] = gaps

```

```python
import pandas as pd

# Build a flat list of records: one row per gap, then pivot to counts
records = []
for station, var_gaps in gap_summary.items():
    for var, gaps in var_gaps.items():
        for g in gaps:
            records.append({
                'station': station,
                'variable': var,
                'severity': g['severity']
            })

gap_df = pd.DataFrame(records)

# Pivot: rows = station+variable, columns = severity, values = count
summary_table = (
    gap_df
    .groupby(['station', 'variable', 'severity'])
    .size()
    .unstack(fill_value=0)
    .reindex(columns=['small', 'medium', 'long'], fill_value=0)  # enforce column order
)

summary_table
```

#### Feature Engineering

```python
import numpy as np

def build_feature_matrix(sdf, target_col):
    """Build a feature matrix for ML-based imputation of target_col."""
    feats = pd.DataFrame(index=sdf.index)

    # Time-based cyclical features
    feats['hour_sin'] = np.sin(2 * np.pi * sdf.index.hour / 24)
    feats['hour_cos'] = np.cos(2 * np.pi * sdf.index.hour / 24)
    feats['month_sin'] = np.sin(2 * np.pi * sdf.index.month / 12)
    feats['month_cos'] = np.cos(2 * np.pi * sdf.index.month / 12)
    feats['doy_sin'] = np.sin(2 * np.pi * sdf.index.dayofyear / 365)
    feats['doy_cos'] = np.cos(2 * np.pi * sdf.index.dayofyear / 365)

    # Other variables as context (excluding target and non-numeric like wd)
    other_cols = [c for c in sdf.columns if c not in [target_col, 'wd']]
    for c in other_cols:
        feats[c] = sdf[c]

    return feats.fillna(feats.median(numeric_only=True))
```

#### Efficient masked-validation function

```python
import os
import numpy as np
import pandas as pd
from scipy.interpolate import PchipInterpolator
from lightgbm import LGBMRegressor
from sklearn.metrics import r2_score
from joblib import Parallel, delayed

def get_valid_windows(series, gap_length, buffer=24, max_windows=15, seed=42):
    """Find non-overlapping windows of real data long enough to mask + test."""
    rng = np.random.default_rng(seed)
    n = len(series)
    valid = series.notna().values
    candidates = []
    i = buffer
    while i < n - gap_length - buffer:
        if valid[i:i+gap_length].all():
            candidates.append(i)
            i += gap_length + buffer  # skip ahead to keep windows non-overlapping
        else:
            i += 1
    if len(candidates) > max_windows:
        candidates = rng.choice(candidates, size=max_windows, replace=False)
    return candidates

def evaluate_gap_length(sdf, target_col, gap_length):
    series = sdf[target_col]
    starts = get_valid_windows(series, gap_length)
    if len(starts) == 0:
        return None

    test_idx = []
    masked_series = series.copy()
    for s in starts:
        idx_range = list(range(s, s + gap_length))
        test_idx.extend(idx_range)
        masked_series.iloc[idx_range] = np.nan

    true_values = series.iloc[test_idx].values

    # --- Interpolation ---
    pred_interp = masked_series.interpolate(method='linear', limit_direction='both').iloc[test_idx].values

    # --- PCHIP ---
    valid_mask = masked_series.notna()
    pchip_func = PchipInterpolator(np.where(valid_mask)[0], masked_series[valid_mask].values)
    pred_pchip = pchip_func(test_idx)

    # --- LGBM only (RF dropped for speed — optimization #2) ---
    X = build_feature_matrix(sdf, target_col)
    y = masked_series
    train_mask = y.notna()

    lgbm = LGBMRegressor(n_estimators=100, learning_rate=0.1, max_depth=6, verbose=-1)
    lgbm.fit(X[train_mask], y[train_mask])
    pred_lgbm = lgbm.predict(X.iloc[test_idx])

    return {
        'Interp_R2': r2_score(true_values, pred_interp),
        'PCHIP_R2': r2_score(true_values, pred_pchip),
        'LGBM_R2': r2_score(true_values, pred_lgbm),
    }
```

Apply to all the features

```python
# --- Optimization #1: trimmed grid (drop redundant lengths that rarely shift the story) ---
gap_lengths_to_test = [1, 3, 6, 12, 24, 48]
target_vars = ['PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'WSPM']

# --- Optimization #3: parallelize across station-variable pairs, with per-station caching ---
os.makedirs('cache', exist_ok=True)

def run_one_station(station, sdf, target_vars, gap_lengths):
    cache_path = f'cache/{station.replace(" ", "_")}_comparison.parquet'
    if os.path.exists(cache_path):
        return pd.read_parquet(cache_path)

    station_results = []
    for var in target_vars:
        for gap_len in gap_lengths:
            res = evaluate_gap_length(sdf, var, gap_len)
            if res is not None:
                res.update({'station': station, 'variable': var, 'gap_hours': gap_len})
                station_results.append(res)

    station_df = pd.DataFrame(station_results)
    station_df.to_parquet(cache_path)  # cache so a crash doesn't lose this station's work
    return station_df

all_station_dfs = Parallel(n_jobs=-1, verbose=5)(
    delayed(run_one_station)(station, sdf, target_vars, gap_lengths_to_test)
    for station, sdf in station_dfs.items()
)

comparison_df = pd.concat(all_station_dfs, ignore_index=True)
comparison_df.head()
```

Determine the best model

```python
methods = ['Interp_R2', 'PCHIP_R2', 'LGBM_R2']

def summarize_best_method(var_data):
    """
    Given rows for one station+variable (sorted by gap_hours), find the best
    method at each gap length, then merge consecutive gap lengths that share
    the same winning method into a single range.
    """
    var_data = var_data.sort_values('gap_hours').reset_index(drop=True)

    # Best method + its R2 value at each individual gap length
    var_data['best_method'] = var_data[methods].idxmax(axis=1)
    var_data['best_r2'] = var_data[methods].max(axis=1)

    # Group consecutive rows where best_method stays the same
    segments = []
    current_method = None
    seg_start = None
    seg_end = None
    seg_r2_values = []

    for _, row in var_data.iterrows():
        if row['best_method'] != current_method:
            # close off the previous segment
            if current_method is not None:
                segments.append({
                    'method': current_method,
                    'start': seg_start,
                    'end': seg_end,
                    'r2': seg_r2_values[-1]  # R2 at the end of this segment's range
                })
            # start a new segment
            current_method = row['best_method']
            seg_start = row['gap_hours']
            seg_r2_values = []

        seg_end = row['gap_hours']
        seg_r2_values.append(row['best_r2'])

    # close the final segment
    if current_method is not None:
        segments.append({
            'method': current_method,
            'start': seg_start,
            'end': seg_end,
            'r2': seg_r2_values[-1]
        })

    return segments

def format_range(seg, is_first, is_last):
    method_name = seg['method'].replace('_R2', '')
    r2 = seg['r2']
    if is_first:
        return f"<{seg['end']}h {method_name} R2={r2:.3f}"
    elif is_last:
        return f">{seg['start']}h {method_name} R2={r2:.3f}"
    else:
        return f"{seg['start']}-{seg['end']}h {method_name} R2={r2:.3f}"

# --- Generate text output for every station + feature ---
for station in comparison_df['station'].unique():
    print(f"\n{station}")
    station_df = comparison_df[comparison_df['station'] == station]

    for var in station_df['variable'].unique():
        var_data = station_df[station_df['variable'] == var]
        segments = summarize_best_method(var_data)

        parts = []
        for i, seg in enumerate(segments):
            is_first = (i == 0)
            is_last = (i == len(segments) - 1)
            parts.append(format_range(seg, is_first, is_last))

        line = f"  {var}: " + " | ".join(parts)
        print(line)
```

```python
def find_best_method_and_cutoff(var_data):
    """
    Given all gap-length rows for one station+variable, find which method
    wins at each gap length, and identify the first gap length where LGBM
    becomes the best-performing method (the ML 'cutoff').
    """
    var_data = var_data.sort_values('gap_hours').reset_index(drop=True)

    methods = ['Interp_R2', 'PCHIP_R2', 'LGBM_R2']
    best_per_row = var_data[methods].idxmax(axis=1)
    var_data = var_data.assign(best_method=best_per_row.values)

    # Find the first gap length where LGBM becomes the best method
    crossover_row = var_data[var_data['best_method'] == 'LGBM_R2']
    cutoff = crossover_row['gap_hours'].min() if not crossover_row.empty else None

    return cutoff, var_data
```

```python
cutoff_summary = []
for (station, var), group in comparison_df.groupby(['station', 'variable']):
    cutoff, _ = find_best_method_and_cutoff(group)
    cutoff_summary.append({'station': station, 'variable': var, 'ml_cutoff_hours': cutoff})

cutoff_df = pd.DataFrame(cutoff_summary)
```

Apply the imputation

```python
def apply_final_imputation(sdf, cutoff_df, station):
    """Apply interpolation for small gaps, ML model for large gaps, per variable."""
    sdf_imputed = sdf.copy()
    station_cutoffs = cutoff_df[cutoff_df['station'] == station].set_index('variable')['ml_cutoff_hours']

    for var in sdf.columns:
        if var == 'wd':  # handle categorical separately if needed
            continue
        if var not in station_cutoffs.index or pd.isna(station_cutoffs.get(var)):
            # No ML crossover found — just interpolate everything
            sdf_imputed[var] = sdf_imputed[var].interpolate(method='linear', limit_direction='both')
            continue

        cutoff_hours = station_cutoffs[var]
        series = sdf_imputed[var]
        is_na = series.isna()
        group_id = (is_na != is_na.shift()).cumsum()

        # Separate small vs large gap groups first
        small_gap_idx = []
        large_gap_groups = []
        for _, grp in series[is_na].groupby(group_id[is_na]):
            if len(grp) <= cutoff_hours:
                small_gap_idx.extend(grp.index)
            else:
                large_gap_groups.append(grp)

        # --- Handle all small gaps at once ---
        if small_gap_idx:
            interpolated = sdf_imputed[var].interpolate(method='linear', limit_direction='both')
            sdf_imputed.loc[small_gap_idx, var] = interpolated.loc[small_gap_idx]

        # --- Train ONE model per variable, reuse for all large gaps ---
        if large_gap_groups:
            X = build_feature_matrix(sdf_imputed, var)
            train_mask = sdf_imputed[var].notna()

            model = LGBMRegressor(n_estimators=200, learning_rate=0.05, verbose=-1)
            model.fit(X[train_mask], sdf_imputed[var][train_mask])

            for grp in large_gap_groups:
                sdf_imputed.loc[grp.index, var] = model.predict(X.loc[grp.index])

    return sdf_imputed

# Apply to all stations — parallelized (optimization #2)
from joblib import Parallel, delayed

imputed_results = Parallel(n_jobs=-1, verbose=5)(
    delayed(apply_final_imputation)(sdf, cutoff_df, station)
    for station, sdf in station_dfs.items()
)

imputed_station_dfs = dict(zip(station_dfs.keys(), imputed_results))
```
